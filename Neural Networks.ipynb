```python
# ============================================
# Introduction to Neural Networks - Assignment
# ============================================

import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------
# 1. Activation Functions
# --------------------------------------------

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)


def relu(x): 
    return np.maximum(0, x)


# --------------------------------------------
# 2. Sample Data
# --------------------------------------------

x1 = 3
x2 = 2

y_actual = 17


# --------------------------------------------
# 3. Initialize Random Weights
# --------------------------------------------

np.random.seed(42)

w1 = np.random.uniform(1, 10)
w2 = np.random.uniform(1, 10)

print("Initial w1:", w1)
print("Initial w2:", w2)


# --------------------------------------------
# 4. Learning Rate and Epochs
# --------------------------------------------

learning_rate = 0.01
epochs = 20


# Lists to store results
errors = []
predictions = []


# --------------------------------------------
# 5. Training
# --------------------------------------------

for epoch in range(epochs):

    # -------- Feed Forward --------

    y_pred = x1 * w1 + x2 * w2


    # -------- Calculate Error --------

    error = (y_actual - y_pred) ** 2


    # Store values
    errors.append(error)
    predictions.append(y_pred)


    # -------- Backpropagation --------

    # Gradient of error with respect to w1
    gradient_w1 = -2 * x1 * (y_actual - y_pred)

    # Gradient of error with respect to w2
    gradient_w2 = -2 * x2 * (y_actual - y_pred)


    # -------- Update Weights --------

    w1 = w1 - learning_rate * gradient_w1
    w2 = w2 - learning_rate * gradient_w2


    # Print progress
    print(
        f"Epoch {epoch + 1}: "
        f"Prediction = {y_pred:.4f}, "
        f"Error = {error:.4f}, "
        f"w1 = {w1:.4f}, "
        f"w2 = {w2:.4f}"
    )


# --------------------------------------------
# 6. Final Result
# --------------------------------------------

print("\nFinal Results")
print("-------------")

print("Final w1:", w1)
print("Final w2:", w2)

print("Actual y:", y_actual)
print("Final predicted y:", predictions[-1])
print("Final error:", errors[-1])


# --------------------------------------------
# 7. Plot Error vs Epoch
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, epochs + 1),
    errors,
    marker='o'
)

plt.xlabel("Epoch")
plt.ylabel("Error")
plt.title("Error Progression Across Epochs")
plt.grid(True)

plt.show()


# --------------------------------------------
# 8. Plot Predicted Y vs Epoch
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, epochs + 1),
    predictions,
    marker='o',
    label="Predicted Y"
)

plt.axhline(
    y=y_actual,
    linestyle='--',
    label="Actual Y = 17"
)

plt.xlabel("Epoch")
plt.ylabel("Predicted Y")
plt.title("Predicted Y Progression Across Epochs")
plt.legend()
plt.grid(True)

plt.show()
```
